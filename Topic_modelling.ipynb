```python
# Function to download file from Google Drive
def download_google_drive_file(file_url):
    file_id = file_url.split('/d/')[1].split('/')[0]
    download_url = f"https://drive.google.com/uc?export=download&id={file_id}"

    session = requests.Session()
    response = session.get(download_url, stream=True)

    # Handle large files with confirmation token if needed
    token = None
    for key, value in response.cookies.items():
        if key.startswith("download_warning"):
            token = value
            break

    if token:
        params = {"id": file_id, "confirm": token}
        response = session.get(download_url, params=params, stream=True)

    response.raise_for_status()
    return response.content
```

```python
# Datasets( Google and trustPilot reviews) data pre-processing function

def preprocess_review(reviews):

    # Handle different input types and NaN values
    if isinstance(reviews, list):
        reviews = ' '.join(str(review) for review in reviews
                         if not (isinstance(review, float) and np.isnan(review)))
    elif isinstance(reviews, float) and np.isnan(reviews):
        return []
    else:
        reviews = str(reviews)

    # Remove emojis
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F1E0-\U0001F1FF"  # flags (iOS)
        "\U00002702-\U000027B0"  # dingbats
        "\U000024C2-\U0001F251"  # enclosed characters
        "]+",
        flags=re.UNICODE
    )
    reviews = emoji_pattern.sub(r'', reviews)

    # Remove unwanted characters, number, nulls and patterns

    reviews = re.sub(r'\d+', '', str(reviews))  # Remove all digits
    reviews = re.sub(r'@[A-Za-z0-9]+', '',reviews)# removed @mentions
    reviews = re.sub(r'#', '',reviews)# removed # symbol
    reviews = re.sub(r'https?:\/\/\s+', '',reviews)# removed the hyperlink
    reviews = re.sub(r'http', '',reviews)
    reviews = re.sub('[^A-Za-z]+', ' ', reviews)
    reviews = re.sub(r'[^\w\s]', '', reviews)
    reviews = re.sub(r'\s+', ' ', reviews).strip()  # Normalize whitespace

    # changing to lower case and Tokenise the reviews.

    tokens = word_tokenize(reviews.lower())


    # Remove stopwords.

    filtered_tokens = [token for token in tokens if token not in stopwords.words('english')]


    # Lemmatise the tokens.

    lemmatizer = WordNetLemmatizer()

    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]

    # Join the tokens back into a string.

    processed_reviews = ' '.join(lemmatized_tokens  )

    return processed_reviews
```

```python
def plot_word_frequencies(review_series, title="Top Words in Reviews", top_n=10, color='skyblue'):

    # Combine all reviews into one string, handling NaN values
    reviews_text = ' '.join(review_series.dropna().astype(str))

    # Remove specific punctuation (., ,'!) using regex
    reviews_text_cleaned = re.sub(r'[.,\'!’]+', '', reviews_text)
    reviews_text_cleaned = re.sub('[^A-Za-z]+', ' ', reviews_text)

    # Tokenize into words and convert to lowercase
    tokens = word_tokenize(reviews_text_cleaned.lower())

    stop_words = set(stopwords.words('english'))
    tokens = [w for w in tokens if not w in stop_words]


    # Remove all words less than two

    tokens = [word for word in tokens if len(word) > 2]

    # Create frequency distribution
    freq_dist = FreqDist(tokens)

    # Get the top N most common words
    top_words = freq_dist.most_common(top_n)

    # Print frequency distribution
    print(f"Frequency distribution for {title} (top {top_n}):")
    for words, freq in top_words:
        print(f"{words}: {freq}")
    print()

    # Unpack words and frequencies for plotting
    words, freqs = zip(*top_words)

    # Create the bar plot
    plt.figure(figsize=(10, 6))
    plt.bar(words, freqs, color=color)
    plt.title(title)
    plt.xlabel('Words')
    plt.ylabel('Frequency')
    plt.xticks(rotation=45, ha='right')  # Rotate labels for readability
    plt.tight_layout()
    plt.show()
```

```python
def word_cloud_display(df, key):
  top_number_of_results = 100
  all_reviews = df[key].str.lower().str.cat(sep=' ')
  all_text = re.sub('[^A-Za-z]+', ' ', all_reviews)

  stop_words = set(stopwords.words('english'))
  word_tokens = word_tokenize(all_text)
  filtered_all_reviews = [w for w in word_tokens if not w in stop_words]

  all_reviews_no_short_words = [word for word in filtered_all_reviews if len(word) > 2]
  all_reviews_no_numbers = [word for word in all_reviews_no_short_words if not word.isnumeric()]
  word_distribution = nltk.FreqDist(all_reviews_no_numbers)
  reviews_word_frequency_distribution_df = pd.DataFrame(word_distribution.most_common(top_number_of_results),
                    columns=['Word', 'Frequency'])

  plt.figure(figsize = (100,100))
  wc = WordCloud(background_color = 'black', max_words = 1000,  max_font_size = 50)
  wc.generate(' '.join(all_reviews_no_numbers))
  plt.imshow(wc)
  plt.axis('off')
```

```python
# Function to get dominant emotion and scores
def get_dominant_emotion(text):
    try:
        # Ensure text is a string and truncate manually if needed
        if not isinstance(text, str):
            text = str(text)
        # Run the classifier
        prediction = classifier(text)[0]
        emotion_scores = {item['label']: item['score'] for item in prediction}
        dominant_emotion = max(emotion_scores, key=emotion_scores.get)
        return dominant_emotion, emotion_scores[dominant_emotion], emotion_scores
    except Exception as e:
        print(f"Error processing text: {text[:50]}... - {str(e)}")
        return "error", 0.0, {"error": 0.0}  # Fallback for errors
```

```python
import pandas as pd
from transformers import pipeline, AutoTokenizer, AutoModelForCausalLM
import torch

# Load the Phi-3.5-mini-instruct model and tokenizer
model_name = "microsoft/phi-3.5-mini-instruct"
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="cuda",           # Automatically map to CUDA device(s)
    torch_dtype="auto",          # Auto-select precision (e.g., float16 on GPU)
    trust_remote_code=True       # Allow custom model code
)

# Set up the text generation pipeline (removed device argument)
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

# Define the prompt to prepend to each review
prompt = "In the following customer review, pick out the main 3 topics. Return them in a numbered list format, with each one on a new line.\n\n"


def extract_topics_from_reviews(review_series, max_length=1000):
    reviews = review_series.dropna().astype(str).tolist()
    all_topics = []

    for i, review in enumerate(reviews):
        input_text = prompt + review
        try:
            # Generate output with truncation explicitly enabled
            output = generator(
                input_text,
                max_length=max_length,
                num_return_sequences=1,
                do_sample=False,
                truncation=True
            )
            generated_text = output[0]['generated_text']
            topics = generated_text[len(input_text):].strip()

            # Debug: Show the raw output
            print(f"\nReview {i + 1}:")
            print(f"Original Review: {review}")
            print("Raw Generated Topics:")
            print(repr(topics))  # Use repr() to see exact output including newlines

            # Robust topic extraction
            if topics:
                # Split by newlines and filter out empty lines
                topic_lines = [line.strip() for line in topics.split('\n') if line.strip()]
                topic_list = []
                for line in topic_lines:
                    # Try to split on '. ' but handle cases where it fails
                    parts = line.split('. ', 1)
                    if len(parts) == 2:
                        topic_list.append(parts[1])
                    else:
                        # If no '. ' found, use the whole line as a topic
                        topic_list.append(line)
                all_topics.extend(topic_list[:3])  # Take up to 3 topics
            else:
                print("No topics generated.")

            print("Extracted Topics:")
            print('\n'.join(topic_list) if topic_list else "None")
            print("-" * 50)
        except Exception as e:
            print(f"Error processing review {i + 1}: {e}")
            print("-" * 50)

    return all_topics
```

```python
# Initialize lemmatizer
lemmatizer = WordNetLemmatizer()

# Preprocessing function for a single review
def preprocess_review(review):
    if pd.isna(review):
        return []
    review = str(review).lower()
    review = re.sub(r'[^\w\s]', '', review)
    tokens = word_tokenize(review)
    tokens = [token for token in tokens if token not in STOPWORDS and len(token) > 2]
    tokens = [lemmatizer.lemmatize(token) for token in tokens]
    return tokens

# Prepare LDA input
def prepare_lda_input(review_series):
    if isinstance(review_series, pd.Series):
        reviews = review_series.dropna().astype(str).tolist()
    elif isinstance(review_series, list):
        reviews = [str(review) for review in review_series if review is not None]
    else:
        raise ValueError("Input must be a pandas Series or a list")

    processed_reviews = [preprocess_review(review) for review in reviews]
    processed_reviews = [doc for doc in processed_reviews if doc]  # Remove empty lists
    dictionary = corpora.Dictionary(processed_reviews)
    dictionary.filter_extremes(no_below=2, no_above=0.5)
    corpus = [dictionary.doc2bow(doc) for doc in processed_reviews]
    return dictionary, corpus, processed_reviews
```
